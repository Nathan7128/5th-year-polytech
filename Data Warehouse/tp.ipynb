### 1. Quel est le classement des produits suivant leur nombre de ventes (Unit Sales) ?  
### Donner un résultat partiel.

```python
nb_ventes_produits_df = (
    tb_sales_fact_1997_data_df
    .groupby("product_id")
    .apply(lambda x: pd.Series({"nb_ventes": x["unit_sales"].sum()}))
    .sort_values(by="nb_ventes", ascending=False)
)

nb_ventes_produits_df
```

### 4. Proposer un programme permettant de retrouver le résultat de la requête MDX 2 (transparent 35 du cours), sans que l'affichage soit forcément identique.

```python
sales_by_months_df = (
    tb_sales_fact_1997_data_df
    .merge(tb_time_by_day_data_df[tb_time_by_day_data_df.the_year == 1997], on="time_id", how="inner")
    .groupby("month_of_year", as_index=False)
    .apply(lambda x: pd.Series({"Unit Sales": x["unit_sales"].sum()}))
    .sort_values("month_of_year")
    .rename({"month_of_year": "Time"}, axis=1)
    .set_index("Time")
)

sales_by_months_df
```

### 5. Proposer un programme permettant de retrouver le résultat de la requête MDX 4 (transparent 39 du cours), sans que l'affichage soit forcément identique

```python
sales_by_months_quarter_df = (
    tb_sales_fact_1997_data_df
    .merge(tb_time_by_day_data_df[tb_time_by_day_data_df.the_year == 1997], on="time_id", how="inner")
    .groupby(["month_of_year", "quarter"], as_index=False)
    .apply(lambda x: pd.Series({"Unit Sales": x["unit_sales"].sum()}))
)

sales_by_quarters_df = (
    sales_by_months_quarter_df
    .groupby("quarter")
    .apply(lambda x: pd.Series({"Unit Sales Quarter": x["Unit Sales"].sum()}))
)

sales_by_months2_df = (
    sales_by_months_quarter_df
    .merge(sales_by_quarters_df, on="quarter")
    .sort_values(["Unit Sales Quarter", "Unit Sales"], ascending=False)
    .set_index("month_of_year")
)

sales_by_months2_df[["Unit Sales"]]
```

### 6. Proposer un programme permettant de retrouver le résultat de la requête MDX 7 (transparent 45 du cours), sans que l'affichage soit forcément identique.

```python
sales_by_months_df
```

```python
evo_sales_by_months_df = sales_by_months_df.copy()
evo_sales_by_months_df["Unit SalesM-1"] = evo_sales_by_months_df["Unit Sales"].shift(1)
evo_sales_by_months_df["Evolution"] = evo_sales_by_months_df["Unit Sales"] > evo_sales_by_months_df["Unit SalesM-1"]
evo_sales_by_months_df["Evolution"].replace({False: "-", True: "+"}, inplace=True)
evo_sales_by_months_df["Evolution"].iloc[0] = None
evo_sales_by_months_df
```

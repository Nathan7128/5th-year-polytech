```python
import pandas as pd
```

```python
tb_time_by_day_data_df = pd.read_csv("polytech2025edd_tp_source/tb_time_by_day_data.csv")
tb_time_by_day_data_df = tb_time_by_day_data_df[["time_id", "the_year", "quarter", "month_of_year", "day_of_month"]]
tb_time_by_day_data_df.to_csv("processed_tp_data/tb_time_by_day_data.csv", index=False)
```

```python
tb_store_data_df = pd.read_csv("polytech2025edd_tp_source/tb_store_data.csv")
tb_store_data_df = tb_store_data_df[["store_id", "store_country", "store_state", "store_city", "store_name"]]
tb_store_data_df.to_csv("processed_tp_data/tb_store_data.csv", index=False)
```

```python
tb_product_data_df = pd.read_csv("polytech2025edd_tp_source/tb_product_data.csv")
tb_product_data_df = tb_product_data_df[["product_class_id", "product_id", "brand_name", "product_name"]]
tb_product_class_data_df = pd.read_csv("polytech2025edd_tp_source/tb_product_class_data.csv")
tb_product_df = pd.merge(tb_product_data_df, tb_product_class_data_df, on="product_class_id")
tb_product_df.to_csv("processed_tp_data/tb_product.csv", index=False)
```

```python
tb_sales_fact_1997_data_df = pd.read_csv("polytech2025edd_tp_source/tb_sales_fact_1997_data.csv")
tb_sales_fact_1997_data_df.drop(["customer_id", "promotion_id"], axis = 1, inplace=True)
tb_sales_fact_1997_data_df.to_csv("processed_tp_data/tb_sales_fact_1997_data.csv", index=False)
```

```python
tb_profit_df = tb_sales_fact_1997_data_df[["time_id", "product_id", "store_id", "store_sales", "store_cost"]]
tb_profit_df["profit_made"] = tb_profit_df["store_sales"] - tb_profit_df["store_cost"]
tb_profit_df.drop(["store_sales", "store_cost"], axis=1, inplace=True)
tb_profit_df.to_csv("processed_tp_data/tb_profit.csv", index=False)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def phi(x, r):
    return x**2 + r * np.maximum(0, 4 - x)

x = np.linspace(-6, 6, 500)

r_values = [1, 2, 3]

plt.figure(figsize=(10,6))
for r in r_values:
    y = phi(x, r)
    plt.plot(x, y, label=f'r = {r}')

plt.title(r"Fonction $\phi(x, r_k) = x^2 + r_k \max(0, 4-x)$")
plt.xlabel('x')
plt.ylabel(r'$\phi(x, r_k)$')
plt.legend()
plt.grid(True)
plt.show()
```

```python
class RandomAgent:
    def __init__(self, env: gym.Env, learning_rate, discount_factor):
        self.env = env

        self.positions_space = np.linspace(-0.2, 0.6, 10)
        self.velocities_space = np.linspace(-0.07, 0.07, 100)

        self.qtable = np.zeros(shape=(len(self.positions_space)*len(self.velocities_space), 3))

        self.learning_rate = learning_rate
        self.discount_factor = discount_factor

    def get_action(self, state):
        position, velocity = state
        id_position = np.abs(self.positions_space - position).argmin()
        id_velocity = np.abs(self.velocities_space - velocity).argmin()
        state = id_position*10 + id_velocity
        return np.argmax(self.qtable[state])
    
    def update_param(self, state, action, reward, next_action, next_state):

        action = int(np.sign(action))
        next_action = int(np.sign(next_action))

        position, velocity = state
        id_position = np.abs(self.positions_space - position).argmin()
        id_velocity = np.abs(self.velocities_space - velocity).argmin()

        next_position, next_velocity = next_state
        id_next_position = np.abs(self.positions_space - next_position).argmin()
        id_next_velocity = np.abs(self.velocities_space - next_velocity).argmin()

        state = id_position*10 + id_velocity
        next_state = id_next_position*10 + id_next_velocity
        
        old_value = self.qtable[state, action]
        next_value = self.qtable[next_state, next_action]
        new_value = self.learning_rate * (reward + self.discount_factor*next_value + old_value)
        self.qtable[state, action] += new_value
```

```python
epsilon = 0.2
n_episodes = 1000
timesteps_per_episode = []
penalties_per_episode = []
learning_rate = 0.1
discount_factor = 0.2
```

```python
env = gym.make('MountainCarContinuous-v0', render_mode="human")
agent = RandomAgent(env=env, learning_rate=learning_rate, discount_factor=discount_factor)
```

```python
position = random.uniform(-0.6, -0.4)
velocity = 0

for i in range (n_episodes):
    state = env.reset(seed=42)[0]

    position, velocity = state

    action = [0.0]

    epochs, penalties, reward, = 0, 0, 0
    done = False

    while not done:
        if random.uniform(0, 1) < epsilon:
            next_action = env.action_space.sample()

        else:
            next_action = agent.get_action(state=state)

        next_state, reward, terminated, truncated, info = env.step(action)
        done = terminated or truncated
        agent.update_param(state=state, action=action, reward=reward, next_action=next_action, next_state=next_state)

        if reward == -1:
            penalties += +1

        state = next_state
        action = [next_action]

        epochs += 1
        penalties += reward

    timesteps_per_episode.append(epochs)
    penalties_per_episode.append(penalties)
```
